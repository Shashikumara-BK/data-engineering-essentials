# What is lit() in PySpark?
lit stands for literal.
It creates a column object with a constant value (a literal).
This is useful when you want to add a new column with a fixed value or use a constant inside expressions.

withColumn("country", lit("India"))
withColumn → used to add a new column or replace an existing column in a DataFrame.
"country" → is the name of the new column you are adding.
lit("India") → lit is short for literal.
It creates a constant column value for every row.
Here, "India" is treated as a constant literal, so every row will get India in the country column.

```python
df.withColumn("salary1",df.salary*10).show()
```

```python
from pyspark.sql.functions import lit
df.withColumn("country",lit("India")).show()
```

```python
df.withColumn("increment",lit(0)).show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col
df.withColumn('salary',col('salary').cast('int')).show()
```

```python
df1 = df.withColumn('salary',col('salary').cast('int'))
```

```python
df1.printSchema()
```

```python
df.withColumnRenamed("loc","location").show()
```
