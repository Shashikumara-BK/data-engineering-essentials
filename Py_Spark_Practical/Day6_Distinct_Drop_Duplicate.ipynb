# Distinct() and DropDuplicates() Functions
# Distinct() : 
Pyspark Distinct() transformation is used to drop or remove the dupliacate rows(all columns) from DataFrame.
# DropDuplicates() : 
It is used to drop rows based on selected (one or multiple) columns.


```python
df = spark.read.csv("/Volumes/development/data/files/orders/emp_1.csv",header = True, inferSchema = True)
df.show()
```

```python
df1= df.distinct()
```

```python
df1.show()
```

```python
df2= df.dropDuplicates(['emp_id','salary'])
df2.show()
```

# Sort() and orderBy()
We can use either sort() or orderBy() function of pyspark DataFrame to sort DataFrame by Descending or Ascending order based on single or multiple columns.

```python
df =spark.read.csv("/Volumes/development/data/files/orders/emp_1.csv",header=True, inferSchema=True)
df.show()
```

```python
df.sort("salary","name").show()
```

```python
df.orderBy("salary").show()
```

```python
df.sort(df.salary.desc()).show()
```

```python
df.orderBy(df.salary.desc(),df.name.asc()).show()
```
