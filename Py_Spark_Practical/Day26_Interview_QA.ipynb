```python
data = [
    ['2024-01-01', 20000],
    ['2024-01-02', 10000],
    ['2024-01-03', 150000],
    ['2024-01-04', 10000],
    ['2024-01-05', 210000]
]

#define column names
columns = ['date', 'sales']

df = spark.createDataFrame(data,columns)
display(df)
```

```python
df.createOrReplaceTempView("sample")
```

```python
%sql
select * from sample
```

```python
%sql
select *, sum(sales) over (order by date asc) as cumm_sales from sample
```

```python
%sql
select *, sum(sales) over (order by date asc) as cumm_sales, lag(sales) over(order by date asc) as prev_sal, lead(sales) over (order by date asc) as next_sal from sample
```

```python
from pyspark.sql.functions import *
from pyspark.sql import Window

window = Window.orderBy('date')
df1 = df.withColumn('cumm_sales',sum(col('sales')).over(window))
df1.withColumn('prev_sal',lag(col('sales')).over(window))
df1.withColumn('next_sal',lag(col('sales')).over(window))
```
