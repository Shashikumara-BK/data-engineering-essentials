# Joins In PySpark
Pyspark Joins is used to combine two DataFrames.
Pyspark join operation combines data from two or more datasets based on common column or key.
It is fundamebtal operation in pyspark and is similar to SQL Joins.
It supports all basic join type operations available in TRADITIOANAL SQL Joins like INNER, LEFT_OUTER, RIGHT_OUTER, LEFT_ANTI, LEFT_SEMI, CROSS, SELF JOIN.
1. INNER_JOIN:
Returns only the ROWS with matching keys in both DataFrames.
2. LEFT_JOIN:
Returns all ROWS from the left DataFrames and matching ROWS from the right dataframe.
3. RIGHT_JOIN:
Returns all ROWS from the right DataFrame and matching ROWS from the left DataFrame.
4. FULL_OUTER_JOIN:
Returns all ROWS from the both DataFrames , including matching and not matching Rows.
5. LEFT_SEMI_JOIN:
Returns all rows from the left DataFrames where there is a match in the right DataFrame.
6. LEFT_ANTI_JOIN:
Returns all rows from the left DataFrame where there is no match in the right DataFrame.


```python
emp_df = spark.read.csv("/Volumes/development/data/files/orders/employee.csv",header=True,inferSchema=True)
display(emp_df)
```

```python
dep_df = spark.read.csv("/Volumes/development/data/files/department.csv",header=True,inferSchema=True)
display(dep_df)
```

```python
emp_df.join(dep_df,emp_df.emp_id==dep_df.user,"inner").display()
```

```python
emp_df.join(dep_df,emp_df.emp_id==dep_df.user,"left").show()
```

```python
emp_df.join(dep_df,emp_df.emp_id==dep_df.user,"right").show()
```

```python
emp_df.join(dep_df,emp_df.emp_id==dep_df.user,"full_outer").show()
```

```python
emp_df.join(dep_df,emp_df.emp_id==dep_df.user,"leftanti").show()
```

```python
emp_df.join(dep_df,emp_df.emp_id==dep_df.user,"leftsemi").show()
```
