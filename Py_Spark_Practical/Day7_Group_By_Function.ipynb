# GroupBy() in Pyspark
It allows you to perform aggregate functions on group of rows , rather than on individual rows, enabling you to summarize data and generate aggregate statistics.
Aggregate Functions
1. count() Return the number of rows for each group.
2. mean() Returns the mean of values for each group.
3. max() Returns the maximum of values for each group.
4. min() Returns the minimum of values for each group.
5. sum() Returns the total for values for each group.
6. avg() Returns the average for values for each group.
7. agg() Using groupBy() agg() function , we can calculate more than one aggregate at a time.
🔑 Remember
agg() needs both function + column
Example functions: sum, avg, min, max, count, collect_list, collect_set, etc.


```python
df = spark.read.csv("/Volumes/development/data/files/orders/emp_1.csv",header=True,inferSchema=True)
df.show()
```

```python
df.groupBy("address").count().show()
```

```python
df.groupBy("salary").sum("salary").show()
```

```python
df.groupBy("address").sum('salary').show()
```

```python
df.groupBy("address","emp_id").sum('salary').show()
```

```python
df.groupBy("address").max('salary').show()
```

```python
df.groupBy("address").min('salary').show()
```

```python
df.groupBy("address").avg('salary').show()
```

```python
df.groupBy("address").mean('salary').show()
```

```python
from pyspark.sql import functions as F
df.groupBy("address").agg(F.sum('salary').alias('total_salary'),
                          F.avg('salary').alias('avg_salary'),
                          F.max("salary").alias('max_salary'),
                          F.min("salary").alias('min_salary')
                          ).show()
```
