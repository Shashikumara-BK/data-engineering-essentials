# fill() and fillna() 
in PySpark are often used for handling missing (null/NaN) values.
Both are aliases of each other, so they work the same.
These two functions are used to replace the null or none on all or selected multiple columns with either zero (0) , empty string , space or any constant literal values.
# 🔹 Syntax:
DataFrame.fillna(value, subset=None)
DataFrame.fill(value, subset=None)

value → what to replace null/NaN with
Can be a single value (applied to all columns)
Or a dictionary {col: value} (different replacements per column)
subset → optional, list of columns to apply replacement

# 🔹 Difference between fill() and fillna()
No real difference — fill() is just an alias of fillna().
You can use either. Most people prefer fillna() because it’s clearer.


```python
df= spark.read.csv("/Volumes/development/data/files/orders/sample.csv",header=True,inferSchema=True)
df.show()
```

```python
df.na.fill("unknown").show()
```

```python
df.fillna("").show()
```

For city (string column): nulls were replaced with "" ✅

For population (numeric column): Spark also tried to replace nulls with "" → but since population is numeric, Spark will leave the nulls unchanged (because "" can’t be cast to an integer).
If you want to replace string columns with "" and numeric columns with 0, you need to pass a dictionary:

```python
df.fillna({"city":"","population":0}).show()
```

```python
df.na.fill("",["city"]).na.fill(0,["population"]).show()
```

# collect()
Pyspark RDD/DataFrame collect() ia an action operation that is used to retrieve all the elements of dataset (from all nodes) to the driver node.
we should use the collect() on smaller dataset usually after filter(),group() etc., Retrieving larger dataset results in OutOfMemory error.

# 🔹 What is collect()?
collect() retrieves the entire DataFrame (or RDD) from the cluster back to the driver as a list of Row objects.
It’s mainly used for:
Debugging
Small datasets
Converting Spark data to native Python objects
# ⚠️ Important:
 On big data, collect() can cause Out of Memory errors because it brings all the data to the driver. For large datasets, use .show(), .take(n), or .limit(n) instead.

# 🔹 Related methods
take(n) → returns first n rows as list of Rows (safer than collect).
show(n) → prints nicely formatted output.
toPandas() → converts DataFrame to Pandas DataFrame (only use for small data).
# 👉 In interviews, 
if they ask about collect(), you should mention:
It brings all data to driver → not recommended on large datasets.
Best for small datasets or debugging.
