# date_format in pyspark

The date_format() function in PySpark is a powerful tool for transforming, formatting date columns and converting date to string within a DataFrame.

This function allows you to convert date and timestamp columns into a specified string format, providing flexibility for various date manipulation tasks.

eg. 2024/04/05 ------------- 2024-04-05
2024-mar-05
05/04/2024


date_format in PySpark is a function from pyspark.sql.functions that lets you format date/timestamp columns into a string with a specific pattern.
# Syntax
date_format(col, format)
col: The column name or column expression containing date or timestamp.
format: A Java-style datetime pattern string (e.g. "yyyy-MM-dd", "dd/MM/yyyy", "yyyy MMM dd").
It returns a string column, not a date.

# Common Patterns
"yyyy-MM-dd" → 2025-09-30
"dd/MM/yyyy" → 30/09/2025
"MMMM" → September
"E" → Tue (day name short)
"EEEE" → Tuesday (day name full)
"HH:mm:ss" → 15:45:00

# ⚡ Tip:
If you want to keep it as a date type (not string), you should use to_date() or date_trunc() instead of date_format.

```python
data = [("2022-03-15", "2022-03-16 12:34:56.789"),
        ("2022-03-01", "2022-03-16 01:23:45.678")]
df = spark.createDataFrame(data, ["date", "timestamp"])
df.show()
```

```python
from pyspark.sql.functions import *
df.select("date",date_format("date","yyyy/mm/dd").alias("date")).show()
```

```python
from pyspark.sql.functions import *
df.select("date",date_format("date",'yyyy-MMM-dd').alias('date_col')).show()
```

```python
from pyspark.sql.functions import *
df.select("date",date_format("date",'yyyy-MMMM-dd').alias('date_col')).show()
```

DATE_SUB
date_sub(start_date, num_days): This function returns a new date by subtracting a specified number of days from a given start date.

DATEDIFF
datediff(end_date, start_date): This function returns the number of days between two dates.

YEAR
year(date): This function returns the year component of a date.

DAYOFYEAR
dayofyear(date): Returns the day of the year of the input date
