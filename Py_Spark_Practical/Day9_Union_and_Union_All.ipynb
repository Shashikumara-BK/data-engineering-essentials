# union()
Returns the combined rows of two DataFrames.
Removes duplicate rows (like SQL UNION).
Requires both DataFrames to have the same schema (same number of columns & same data types).
# unionAll()
Returns the combined rows of two DataFrames.
Keeps duplicates (like SQL UNION ALL).
Same schema requirement applies.
# What’s the difference between union and unionAll?
→ Say: "union() removes duplicates, while unionAll() keeps duplicates. But in modern PySpark, unionAll() is deprecated — instead, union() behaves like UNION ALL, and we add .distinct() if we want SQL UNION behavior."

```python
import pyspark
from pyspark.sql import SparkSession
data1= [("James","sales","NY",90000,34,10000),("Michael","sales","NY",86000,56,20000),("Robert","sales","CA",81000,30,23000),("Maria","Finance","CA",90000,24,23000)]

columns1= ["employee_name","department","state","salary","age","bonus"]

df= spark.createDataFrame(data=data1,schema=columns1)
df.show()
```

```python
data2= [("James","sales","NY",90000,34,10000),("Maria","Finance","CA",90000,24,23000),("Jen","Finance","NY",79000,53,15000),("Jeff","Marketing","CA",80000,25,18000),("Kumar","Marketing","NY",91000,50,21000)]
                                                
columns2= ["employee_name","department","state","salary","age","bonus"]
df2= spark.createDataFrame(data=data2,schema=columns2)
df2.show()
```

```python
df.printSchema()
```

```python
df2.printSchema()
```

```python
df.union(df2).show()
```

```python
df.unionAll(df2).show()
```

```python
df.union(df2).distinct().show()
```
