# Databricks Project

## Adidas sales data Analysis

The objective of this project is to analyze the Adidas sales database for the year 2020 and 2021 and identify key insights to help improve sales performance and optimize business strategies.

By analyzing the sales data, we aim to understand factors influencing sales, identify trends, and uncover opportunities for growth. The analysis will be conducted using databricks Notebook to provide an interactive and insightful dashboard.

### Business Metrics requirements

- Total Sales, Total Profit, Average Price per Unit, and Total Units Sold
- Total sales by month
- Total sales by state
- total sales by region
- Total sales by product
- Total sales by retailer
- Units Sold by Product Category and Gender Type
- Top Performing Cities by Profit

```python
df = spark.read.csv("/Volumes/development/data/files/orders/Adidas US Sales Datasets.csv",header = True, inferSchema = True)
df.show()
```

```python
df.createOrReplaceTempView("sales_analysis")
```

```python
%sql
select * from sales_analysis
```

```python
%sql
-- Total Sales, Total Profit, Average Price per Unit, and Total Units Sold
select sum(try_cast(`Total Sales` as double)) as total_sales,
sum(try_cast(`Operating Profit` as double)) as total_profit,
avg(try_cast(`Price per Unit`as double)) as avg_price,
sum(try_cast(`Units Sold`as double)) as total_units
 from sales_analysis
```

```python
%sql
select sum(try_cast(`total sales` as double)) as total_sales,
date_format(`Invoice Date`, 'MMM') as month
from sales_analysis
group by 2
order by month
```

```python
%sql
select sum(try_cast(`total sales` as double)) as total_sales,
State
from sales_analysis
group by 2
order by 1 desc limit 5
```

```python
%sql
select sum(try_cast(`total sales` as double)) as total_sales,
region
from sales_analysis
group by 2
```

```python
%sql
select sum(try_cast(`total sales` as double)) as total_sales,
Product
from sales_analysis
group by 2
order by 1 desc limit 5
```

```python
%sql
select sum(try_cast(`total sales` as double)) as total_sales,
retailer
from sales_analysis
group by 2
order by 1 desc limit 5
```

```python
%sql
select sum(try_cast(`Units Sold` as double)) as total_unit_sold,
`sales method`
from sales_analysis
group by 2
```

```python
%sql
select sum(try_cast(`operating profit` as double)) as total_profit,
city
from sales_analysis
group by 2
order by 1 desc limit 5
```
