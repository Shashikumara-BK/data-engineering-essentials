```python
INTERVIEW QUESTION

WE HAVE 2 TABLE EMPLOYEE AND DEPARTMENT TABLE IS GIVEN

WE HAVE TO FIND THE HIGHEST SALARY BASED ON EACH DEPARTMENT NAME

WE HAVE TO FIND THE EMPLOYEE WHO IS GETTING HIGHEST SALARY BASED ON EACH DEPARTMENT NAME

WE HAVE TO FIND THE LOWEST SALARY BASED ON EACH DEPARTMENT NAME

WE HAVE TO FIND THE EMPLOYEE WHO IS GETTING LOWEST SALARY BASED ON EACH DEPARTMENT NAME

SOLVE USING PYSPARK AND SPARK SQL
```

```python
emp_data = [
('Manish', 1, 75000),
('Raghav', 1, 85000),
('surya', 1, 80000),
('virat', 2, 70000),
('rohit', 2, 75000),
('jadeja', 3, 80000),
('anil', 3, 55000),
('sachin', 3, 55000),
('zahir', 4, 60000),
('bumrah', 4, 65000)
]
schema = ["emp_name", "dept_id", "salary"]

dept_data = [
(1, 'DATA ENGINEER'),
(2, 'SALES'),
(3, 'SOFTWARE'),
(4, 'HR')
]
schema1 = ["dept_id", "dept_name"]

emp_df = spark.createDataFrame(emp_data, schema)
dept_df = spark.createDataFrame(dept_data, schema1)

```

```python
from pyspark.sql.functions import *
df = emp_df.join(dept_df,emp_df.dept_id == dept_df.dept_id,"left")
df1 = df.groupBy('dept_name').agg(max("salary").alias("max_sal"))
```

```python
df1.show()
```

```python
df2= df.join(df1,df.dept_name==df1.dept_name,"inner")
```

```python
df2.filter(col("salary")==col("max_sal")).show()
```
