# select function in pyspark
select() function is used to select single , multiple , column by index, all columns from the list and the nested columns from a dataframe.

```python
df = spark.read.csv('/Volumes/development/data/files/orders/emp_1.csv',header=True,inferSchema=True)
```

```python
df.show()
```

```python
df.select("emp_id","name","salary","address").show()
```

```python
df.select(df.emp_id,df.name,df.salary,df.address).show()
```

```python
df.select(df.columns[:4]).show()
```
