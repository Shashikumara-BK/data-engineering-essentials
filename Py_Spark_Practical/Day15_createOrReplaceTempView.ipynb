# createOrReplaceTempView()

- Temporary views are virtual tables
- They are not stored in the catalog and are only visible to the current session
- They are automatically dropped when the session ends
- Temporary views allow you to reuse the result of a complex query or a DataFrame without creating a permanent table.

- In PySpark, createOrReplaceTempView() is used to register a DataFrame as a temporary SQL view, so that you can run SQL queries directly on it using Spark SQL.
✅ Syntax:
df.createOrReplaceTempView("view_name")
df → Your DataFrame
"view_name" → The name you give to the temporary SQL table (accessible only within the Spark session)

- 🔑 Notes:
- createOrReplaceTempView("name") → Creates a view that exists only during the current Spark session.
- If you run it again with the same name, it will replace the existing view.
- For cross-session persistence, you’d use createOrReplaceGlobalTempView() instead.

```python
data = [(1,'manish',10000),(2,'suresh',20000),(3,'ramesh',30000)]
Columns = ['id','name','salary']

df = spark.createDataFrame(data,Columns)
display(df)
```

```python
df.createOrReplaceTempView('employee')
```

```python
%sql
select * from employee
```

```python
%sql
select max(salary) FROM employee
```

```python
%sql
select upper(name),salary from employee
```
