🔹 What is explode?

In PySpark, explode is a function from pyspark.sql.functions that takes an array or map column and creates multiple rows for each element in the array (or each key-value pair in a map).

It’s very useful when you have nested/array data (common in JSON, logs, semi-structured data).

explode_outer (safer version)

explode will **drop rows if the array/map is null or empty`.

explode_outer keeps the row but returns null.

✅ Summary

explode(array) → creates a new row for each element.

explode(map) → creates a new row for each key-value pair.

explode_outer → keeps null/empty arrays instead of dropping them.

```python
data = [
    ("Manish", ["Java", "Scala", "C++"]),
    ("rahul", ["Spark", "Java", "C++", "Spark", "Java"]),
    ("shyam", ["CSharp", "VB", "Spark", "Python"])
]
columns = ["name", "language"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
from pyspark.sql.functions import *
df.select("name",explode(col("language"))).show()
```

```python
df.createOrReplaceTempView("sample")
```

```python
%sql
select * from sample
```

```python
%sql
select name, explode(language) from sample
```
