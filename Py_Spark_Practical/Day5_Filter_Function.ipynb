# Filter() function
Pyspark Filter() is used to create a new dataframeby filtering the elements from an exeisting Dataframe based on the given condition.
If you have a background in SQL , you can opt to use where() function instead of filter(). Both works identically.
1. Create Dataframe 
2. Filter data based on condition
3. Filter data based on or(|) and (&) condition.
4. Filter data based on startwith , endswith condition
5. Filter data based on expression.

```python
df = spark.read.csv('/Volumes/development/data/files/orders/emp_1.csv', header=True, inferSchema=True)
display(df)
```

```python
from pyspark.sql.functions import col
df.filter(col("address")=="india").show()
#or
df.filter("address='india'").show()
```

```python
df.filter(df.address!='india').show()
```

```python
df.filter((df.name=='manish')&(df.salary=='10000')).show()
```

```python
df.filter((col('name') =='manish')& (col('address') =='india')).show()
```

```python
df.filter((df.name=='manish')|(df.salary=='10000')).show()
```

```python
df.filter((col('name') =='manish')|(col('address') =='india')).show()
```

```python
df.show()
```

```python
df.filter(df.name.startswith('r')).show()
```

```python
df.filter(df.name.endswith('l')).show()
```

```python
df.filter(df.name.contains('h')).show()
```

```python
df.filter(df.name.isNull()).show()
```

```python
df.show()
```

```python
df.filter(df.name.like("%ul%")).show()
```
