```python
INTERVIEW QUESTION

Group multiple rows into single

SOLVE USING PYSPARK AND SPARK SQL
```

```python
data = [
    (1, 'Manish', 'Mobile'),
    (1, 'Manish', 'Washing Mavhine'),
    (2, 'Rahul', 'Car'),
    (2, 'Rahul', 'mobile'),
    (2, 'Rahul', 'scooty'),
    (3, 'Monu', 'Scooty')
]
schema = ["Customer_ID", "Customer_Name", "Purchase"]

df = spark.createDataFrame(data,schema)
display(df)
```

```python
df.createOrReplaceTempView('sample')
```

```python
%sql
select * from sample
```

```python
%sql
select Customer_ID,Customer_Name, collect_set(Purchase) as purchaseNew from sample
group by Customer_ID,Customer_Name
```

```python
from pyspark.sql.functions import collect_set
df.groupBy("Customer_ID", "Customer_Name").agg(collect_set("Purchase")).show()
```
