```python
data = [ ('Manish', '{"street": "123 St", "city": "Delhi"}'), ('Ram', '{"street": "456 St", "city": "Mumbai"}') ]
schema = ['name', 'address']

df = spark.createDataFrame(data, schema)
```

```python
df.createOrReplaceTempView('sample')
```

```python
%sql
select * from sample
```

```python
%sql
with cte(
select name,address,
from_json(address, 'street string, city string') as new_address
from sample)

select name, address, new_address,new_address.street as street,new_address.city as city from cte
```

```python
from pyspark.sql.functions import *

df1=df.withColumn('new_address',from_json(col("address"),'street string, city string'))
```

```python
display(df1.select("name",'address','new_address',col("new_address").street.alias('street'),col("new_address").city.alias('city')))
```
