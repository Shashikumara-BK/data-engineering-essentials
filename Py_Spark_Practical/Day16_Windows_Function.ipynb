# Window functions 

- PySpark Window functions are used to calculate results, such as the rank, row number, etc
- row_number() Returns a sequential number starting from 1 within a window partition
- rank() Returns the rank of rows within a window partition, with gaps.
- dense_rank() Returns the rank of rows within a window partition without any gaps. Where as Rank() returns rank with gaps.
- lag In PySpark, the lag() function retrieves the value of a column from a preceding row within the same window
- lead() function in PySpark retrieves the value of a column from a succeeding row within the same window
- select row_number() over(partition by id order by salary desc) rn from table

- 
- In PySpark, Window functions are powerful operations that allow you to perform calculations across a set of rows related to the current row, without collapsing them into a single output like groupBy does.
- They are useful for things like ranking, row numbers, running totals, moving averages, lag/lead operations, cumulative sums, etc.
- 
- 🔹 Define a Window
- A window specification defines:
- Partition → like SQL PARTITION BY (groups of data)
- Order → like SQL ORDER BY
- Frame → rows/range relative to current row (optional)

Key Points:

rank() vs row_number()

rank() → assigns the same rank to rows with tie values. Next rank will skip numbers (1,1,3…).

row_number() → always gives a unique sequential number (1,2,3…), even for ties.

dense_rank() → assigns the same rank for ties, but does not skip numbers (1,1,2…).

Always import the function from pyspark.sql.functions.

```python

data = [(1,'manish','india',10000),
      (2,'rani','india',50000),
      (3,'sunny','UK',5000),
      (4,'sohan','UK',25000),
      (5,'mona','india',10000)]

column = ['id','name','country','salary']
df = spark.createDataFrame(data,column)
```

row_number
→ always gives a unique sequential number (1,2,3…), even for ties.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,rank
window = Window.partitionBy('country').orderBy('salary')
df.withColumn("rn",row_number().over(window)).show()
```

rank() → assigns the same rank to rows with tie values. Next rank will skip numbers (1,1,3…).

```python
from pyspark.sql.window import Window
window = Window.partitionBy('country').orderBy('salary')
df.withColumn("rn",rank().over(window)).show()
```

dense_rank() → assigns the same rank for ties, but does not skip numbers (1,1,2…).

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import dense_rank
window = Window.partitionBy('country').orderBy('salary')
df.withColumn("rn",dense_rank().over(window)).show()
```

✅ Key points:

Make sure df is a valid DataFrame (not None).

Don’t assign df = df.show() because .show() returns None.

Import all window functions you plan to use (row_number, rank, dense_rank) from pyspark.sql.functions.

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,rank,col
window = Window.partitionBy('country').orderBy(col('salary').desc())
df.withColumn("rn",row_number().over(window)).show()
```
