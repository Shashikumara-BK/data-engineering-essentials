StructType and StructField
Pyspark provides StructType class from pyspark.sql.types to define the structure of the DataFrame.
StructType:
Represents a schema , which is a collection of structfiled objects.
A structType is essentially a list of fileds , each with a name and data type, definingthe structure of the DataFrame

StructField
Represents a filed in the schema, containing metadata such as the name, data type, and nullable status of the field.
Each StructField object defines a single column in the DataFrame, specifying its name and the type of data it holds.

The StructFieled  class is also part of pyspark.sql.types


# 🔹 Why use them?
When you create a DataFrame, Spark can:
Infer schema automatically (based on data)
Or you can define schema explicitly using StructType & StructField → recommended for production (because it’s faster, avoids wrong type inference).

# 🔹 StructType
Represents the schema of the DataFrame (like a table structure).
It is a collection (list) of StructField objects.
# 🔹 StructField
Represents a column inside the schema.
Has 3 main parts:
Column name (str)
Data type (StringType, IntegerType, etc.)
Nullable (True/False → whether null values are allowed)

# 🔹 StructType and StructField Overview
StructType: Defines the schema (like a table structure). It’s a collection of StructFields.
StructField: Defines each column (name, data type, nullable).
This is used when:
Reading raw data (CSV, JSON, Parquet) and you want to enforce a schema.
Creating a DataFrame manually with RDDs.
Making sure Spark does not infer schema incorrectly.

# 🔹 Interview Tips
Say: “I used StructType and StructField to define explicit schemas while creating DataFrames, especially when reading JSON/CSV files where schema inference may cause issues. I also worked with nested StructType for hierarchical data.”
Example use case: Reading JSON with nested structures, enforcing schema for consistency.

```python
from pyspark.sql.types import StructType, StructField, DoubleType, IntegerType, StringType

Data = [(1,"manish","USA"),(2,"mani","USA"),(3,"nish","USA")]

schema = StructType([
    StructField("id",IntegerType()),
    StructField("name",StringType()),
    StructField("location",StringType())
    ])

df= spark.createDataFrame(Data,schema)
df.show()
df.printSchema()
```
